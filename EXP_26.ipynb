```python
import hashlib
import pandas as pd

files = {
    "file1.exe": b"virus sample",
    "file2.exe": b"normal application",
    "file3.exe": b"trojan code"
}

known_malware = []

rows = []

for name, data in files.items():

    md5 = hashlib.md5(data).hexdigest()
    sha1 = hashlib.sha1(data).hexdigest()
    sha256 = hashlib.sha256(data).hexdigest()

    if "virus" in data.decode() or "trojan" in data.decode():
        known_malware.append(sha256)

    rows.append([name, md5, sha1, sha256])

df = pd.DataFrame(rows,
                  columns=["File","MD5","SHA1","SHA256"])

print(df)

print("\nDetection Results")

for _, row in df.iterrows():
    if row["SHA256"] in known_malware:
        print(row["File"], "-> Malware")
    else:
        print(row["File"], "-> Clean")

```
